```python
import pandas as pd
import numpy as np
```

```python
Fruit={'fruits':['apple','orange','banana','orange','guava','apple']}
df=pd.DataFrame(Fruit)
```

```python
unique_value=df["fruits"].unique()
for i in unique_value:
    df[i]=[1 if x==i else 0 for x in df['fruits']]
df.drop('fruits',axis=1,inplace=True)
print(df)
```
