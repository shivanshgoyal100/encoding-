```python
import pandas as pd
import numpy as np
data = {
    'Area':[100,59,np.nan,82,np.nan,68,79],
    'Price':[100000,np.nan,54855,91515,86500,np.nan,71054]
}
df=pd.DataFrame(data)
imputed_data={}
for column in df.columns:
    value = np.array(df[column])
    mean_value = np.nanmean(value)
    imputed_col = {}
    index = 1
    for val in value:
        if np.isnan(val):
            imputed_col[index] = mean_value
        else:
            imputed_col[index] = val
        index += 1

    imputed_data[column] = imputed_col
df_imputed = pd.DataFrame(imputed_data)
print("IMPUTATION with Mean strategy\n")
print(df_imputed)



```
